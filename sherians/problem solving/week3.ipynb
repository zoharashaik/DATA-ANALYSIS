Write a generator that yields even numbers up to N

```python
def even_numbers(n):

    for i in range(2, n + 1):

        if i % 2 == 0:

            yield i

for  num  in even_numbers(10):
    print(num)

```

Handle multiple exceptions in a single try block.

```python
try:
    
    num1 = int(input("Enter first  number:"))
    num2 = int(input("Enter second number:"))

    result = num1 / num2

    print(result)

except ValueError:
    print("Numbers  only please")
except ZeroDivisionError:
    print("You cannot enter second number as zero")
except Exception as err:
    print(f"unexpected error occured as {err}")
finally:
    print("Execution completed")
```

Write a function that returns a new string made of every third character of the original string.

```python
def third_character():
    a = input("Enter string:")
    result = ""
    for i in range(0, len(a), 3):
        result += a[i]

    return result
third_character()
```

```python
def every_third_character():
    a = input("Enter string:")
    print(a[::3])
every_third_character()
```
