Write a function that returns the factorial of a number.

```python
def factorial():
    n = int(input("Enter any number:"))
    fact = 1
    for i in range(1,n+1):
        fact =  fact * i
    print(f"Factorial of a number  is {fact}")
factorial()
```

Create a function that checks whether a string is a palindrome.

```python
def pallindrome():
    a = input("Enter string:")
    
    rev = ""
    for i in range(len(a)-1,-1,-1):
        rev = rev + a[i]
    if rev == a:
        print("string is pallindrome")
    else:
        print("not a pallindrome")
pallindrome()
```

Write a function to remove all punctuation from a string.

```python
import string
def remove_punctuation():
    result = ""
    data = input("Enter string:")
    for char in data:
        if char not in string.punctuation:
            result = result + char
    print(result)
remove_punctuation()    
```

Write a function to flatten a nested list.

```python
def flatten_list(data):
    flat = []
    for sublist in data:
        for item in sublist:
            flat.append(item)
    print(f"flattend list : {flat}")
flatten_list([[1,2],[3,4],[5,6]])
```

Write a function to find the missing number from a list of 1 to N. 

```python
# def missing_number(numbers, n):

#     for i in range(1, n+1):
#         if i not in numbers:
#             return i

# data = [1,2,3,5]
# print(missing_number(data, 5))

def missing_numbers(numbers, n):
    total_sum = n * (n + 1) // 2
    list_sum = sum(numbers)
    missing = total_sum - list_sum
    return missing

data = [1,2,3,5]
result  = missing_numbers(data, 5)
print(result)

```

Write a function to merge two dictionaries and handle key collisions by summing values.

```python
def merge_dict():
    d1 = {"a":100, "b":200}
    d2 = {"b":300, "c":400}
    result = d1.copy()
    for key, value in d2.items():
        if key in result:
            result[key] = result[key] + value
        else:
            result[key] = value
    print(result)
merge_dict()
```
