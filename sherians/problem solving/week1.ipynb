Write a program to calculate the factorial of a number using a loop.

```python
num = int(input("Enter a number:"))
factorial = 1
if num < 0:
    print("Factorial doesnt exist for negative numbers")
else:
    for i in range(1,num+1):
        factorial *= i
    print(f"factorial of a given number is:{factorial}")
```

```python
num = input("Enter your number:")
reverse = ""

for i in num:
    reverse = i + reverse

if num == reverse:
    print("your number is palindrome")
else:
    print("not a palindromic number")
```

Create a number guessing game. 

```python
import random
num = random.randint(1,11)
tries = 0
while True:
    guess = int(input("Enter number you guessed:"))

    if num == guess:
        tries += 1
        print(f"you are right you guessed the number in {tries} tries")
        break
    
    elif num < guess:
        print("Go a little lower")
        tries += 1

    elif num > guess:
        print("Go a little higher")
        tries += 1

    else:
        tries += 1
        print("sorry you are wrong")
```

Check if a given year is a leap year or not.

```python
year = int(input("Enter year:"))
if year % 100 == 0 and year % 400 == 0:
    print("Its a leap year")
elif year % 100 != 0 and year % 4 == 0:
    print("Its a leap year")
else:
    print("Its a normal year")
```
