Create a number guessing game. 

```python
import random
num = random.randint(1,11)
tries = 0
while True:
    guess = int(input("Enter number you guessed:"))

    if num == guess:
        tries += 1
        print(f"you are right you guessed the number in {tries} tries")
        break
    
    elif num < guess:
        print("Go a little lower")
        tries += 1

    elif num > guess:
        print("Go a little higher")
        tries += 1

    else:
        tries += 1
        print("sorry you are wrong")
```

Write a program to find the GCD of two numbers. 

```python
a = int(input("Enter your first number:"))
b =  int(input("Enter your second number:"))

small = min(a, b)

gcd = 1

for i in range(1,small + 1):

    if a % i == 0 and b % i == 0:
        gcd = i
print(f"your gcd is {gcd}")
    
```

 Write a program to calculate the sum of digits of a number. 


```python
num = int(input("Enter a number: "))

# Variable to store sum
total = 0

# Loop until number becomes 0
while num > 0:

    # Extracting last digit
    digit = num % 10

    # Adding digit to total
    total = total + digit

    # Removing last digit
    num = num // 10

# Displaying result
print("Sum of digits is:", total)
```

 Create a program to print all Armstrong numbers between 1 to 1000.

```python
for num in range(1,1001):

    original = num

    digits = len(str(num))

    total = 0

    while num > 0:
        digit = num % 10

        total = total + (digit ** digits)

        num = num // 10

        

    if total == original:
        print(f"{original}", end = " ")
```
