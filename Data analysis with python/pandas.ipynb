```python
# read the data from csv file

dff = pd.read_csv('sales_data.csv')
dff.head(5)
```

```python
# to read only bottom last 5 records from csv file

dff.tail(5)
```

```python
# Display the data types of each column
print("Data types:\n", dff.dtypes)

# Describe the DataFrame
print("Statistical summary:\n", dff.describe())
```

```python
# Describe the DataFrame
print("Statistical summary:\n", dff.describe())
```

```python
dff.describe()
```

```python
# to c the datatypes 
dff.dtypes # obj is ntg but string
```

```python
# Handling missing values
dff.isnull().any() # where there is true there ll b null values
```

```python
dff.isnull().sum() # gonna sum the null values
```

```python
# to fill the null places
dff.fillna(0)
```

```python
# filling the missing values wth the mean of the column
dff['Sales_fillNA']=dff['Sales'].fillna(dff['Sales'].mean())
dff
```

```python
# Renaming the columns
dff=dff.rename(columns={'Date':'Sales_date'})
dff
```

```python
# changing data types
# converting the value from float type to int
dff['Value_new']=dff['Value'].fillna(dff['Value'].mean()).astype(int)
dff.head()  
```

```python
# applying a function to a value column
dff['New value']=dff['Value'].apply(lambda x:x*2)
dff.head()
```

```python
# Data Aggregating and grouping
grouped_mean = dff.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum = dff.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum)
```

```python
# Aggregate multiple functions
grouped_agg = dff.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
# Merging and joining dataframes
# creating sample dataframes
df1 = pd.DataFrame({'Key' : ['A','B','C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'Key' : ['A','B','D'], 'Value2': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
# Merge the dataframes on key column
merge = pd.merge(df1,df2,on='Key',how = 'inner')
merge
```

```python
merge = pd.merge(df1,df2,on='Key',how = 'outer')
merge
```

```python
merge = pd.merge(df1,df2,on='Key',how = 'left')
merge
```

```python
merge = pd.merge(df1,df2,on='Key',how = 'right')
merge
```
