```python
# Q22. Create a program to print all Armstrong numbers between 1 to 1000.

# 3³ + 7³ + 0³
# = 27 + 343 + 0
# = 370
# 370 = 370 Armstrong num

for num in range(1,1001):

    original = num

    digits = len(str(num))

    total = 0

    while num > 0:

        digit = num % 10

        total = total + (digit ** digits)

        num = num // 10

    if total == original:
        print(f"{original}",end = " ")
```

```python
# 29. Write a function to merge two dictionaries and handle key collisions by summing values.

def merge_dict():
    d1 = {"a" : 200 , "b" : 200 , "c" : 500}
    d2 = {"b" : 200 , "d" : 600}

    dict = d1.copy()

    for key,value in d2.items():
        if key in dict:
            dict[key] = dict[key] + value
        else:
            dict[key] = value
    print(dict) 
merge_dict()
```

```python
# 30. Create a function to find unique elements present in only one of two lists.

def unique_elements(l1,l2):
    result = []

    for i in l1:
        if i not in l2:
            result.append(i)

    for i in l2:
        if i not in l1:
            result.append(i)
    return result

l1 = [1,2,3,4,5,6]
l2 = [4,6,5,7,8,9,46,45]

print("unique_elements", unique_elements(l1,l2))
```

```python
# Q19. Write a script to encode a string using Caesar cipher (shift = 3). 

a = input("Enter String:")
Encoded_string = ""

for char in a:
    if char.isupper():
        Encoded_string += chr((ord(char) - ord('A') + 3) % 26 + ord('A'))
    elif char.islower():
        Encoded_string += chr((ord(char) - ord('a') + 3) % 26 + ord('a'))
    else:
        Encoded_string +=  char
print(f"original string {a} after encoding {Encoded_string}")
```

```python
# Q26. Write a program to find all palindromic substrings in a string.

s = input("Enter string:")
palindromes = []

for i in range(len(s)):
    for j in range(i + 1,len(s) + 1):
        sub = s[i:j]

        if sub == sub[::-1] and sub not in palindromes:
            palindromes.append(sub)
print(f"Palindromic substrings {palindromes}")
```

```python
# Q27. Write a function that compresses a string using run-length encoding. 

def RLE():
    s = input("Enter string:")
    result = []
    count = 1

    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result += s[i-1] + str(count)
            count = 1

    if len(s) > 0:
        result += s[-1] +str(count)

    return result
RLE()
```

```python
# Q42. Create a function that validates an email address format.

import re

def validate_email():
    email = input("Enter email adddress: ")

    pattern = r'^[a-zA-Z0-9._%+-]+@[a-zA-z0-9.-]+\.[a-zA-Z]{2,}$'

    if re.fullmatch(pattern, email):
        return "valid email format"
    else:
        return "Invalid email format"
validate_email()
```

```python
# Q46. Write a program that converts a sentence to Pig Latin. 

a = input("Enter sentence : ")
words = a.split()
result = []

for word in words:
    if word[0].lower() in "aeiou":
        new_word = word + "way"
    else:
        new_word = word[1:] + word[0] + "ay"

    result.append(new_word)

print(f"Pig Latin : ", "".join(result))
```

```python
# Q47. Write a script that finds the longest sentence in a paragraph.

p = input("Enter paragraph : ")

sentences = p.split(".")
sentences = [s.strip() for s in sentences if s.strip()]
longest = max(sentences, key=lambda s: len(s.split()))

print(f"longest sentence : {longest}")
```
